```python
#Task 10
ecom["Revenue"]=(ecom["Quantity"])*(ecom["Unit_Price"])
```

```python
conditions=[
    ecom["Revenue"]>=5000,
    ecom["Revenue"]>=2000
]

choices=[
    0.15,
    0.10
]
ecom["Discount_Rate"]=np.select(
    conditions,
    choices,
    default=0.05
)
```

```python
ecom["Discount_Amount"]=(
    ecom["Revenue"]*
    ecom["Discount_Rate"]
)
```

```python
ecom["Final_revenue"]=(
    ecom["Revenue"]*
    ecom["Discount_Amount"]
)
```

```python
ecom["Customer_Status"]=np.where(
    ecom["Rating"] >=4,
    "Satisfied",
    "Needs_Attentions"
)
```

```python
result=ecom[
    (ecom["Category"] == "Electronics")&
    (ecom["Final_revenue"] > 20000) &
    (ecom["Rating"] >= 4) &
    (ecom["Payment_Mode"] == "UPI")
    
]
```

```python
result=result.sort_values(
    "Final_revenue",
    ascending=False
)
```

```python
result = result[
    [
        "Order_ID",
        "Customer",
        "City",
        "Product",
        "Quantity",
        "Revenue",
        "Discount_Amount",
        "Final_revenue",
        "Rating"
    ]
]
```

```python
#Order label

ecom["Order_Label"]=(
    ecom["Order_ID"]
    + " - "
    +ecom["Product"]
    + " - $"
    +ecom["Revenue"].astype(int).astype(str)
)
```

```python
print("\nQUALIFYING ORDERS")
print(result.to_string(index=False))

print("\nORDER LABELS")
print(
    ecom[
        ["Order_ID", "Order_Label"]
    ].to_string(index=False)
)
```
